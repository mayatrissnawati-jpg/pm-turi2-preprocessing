```python
import pandas as pd
import numpy as np

# Dataset karyawan (sengaja mengandung masalah kualitas)
data = {
    'usia'      : [25, 32, np.nan, 45, 28, 51, 38, np.nan, 29, 41],
    'pendapatan': [4_000_000, 7_500_000, 5_200_000, 12_000_000, 4_800_000,
                   15_000_000, 8_100_000, 6_300_000, np.nan, 9_900_000],
    'pendidikan': ['SMA','S1','SMA','S2','SMA','S2','S1','S1','SMA','S2'],  # ordinal
    'kota'      : ['Bandung','Jakarta','Bandung','Surabaya','Jakarta',
                   'Surabaya','Jakarta','Bandung','Bandung','Jakarta'],    # nominal
    'membeli'   : ['Tidak','Ya','Tidak','Ya','Tidak','Ya','Ya','Tidak','Tidak','Ya']  # target
}
df = pd.DataFrame(data)
df
```

```python
# Langkah 1 Menangani nilai hilang (imputasi median/modus)
df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())
df
```

```python
# Langkah 2
X = df.drop(columns=['membeli'])
y = df['membeli']
print(X)
print(y)
```

```python
# langkah 3
X['pendidikan'] = X['pendidikan'].map({'SMA':0,'S1':1,'S2':2}) # ordinal
X = pd.get_dummies(X, columns=['kota'], dtype=int) # nominal
y = y.map({'Tidak':0,'Ya':1})

print(X)
print (y)
```

```python
# langkah 4
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
X, y, test_size=0.3, random_state=42)

print("Jumlah data X_train:", len(X_train))
print("Jumlah data X_test :", len(X_test))
print("Jumlah data y_train:", len(y_train))
print("Jumlah data y_test :", len(y_test))
```

```python
# Langkah 5 Membagi data (SEBELUM scaling)
from sklearn.preprocessing import StandardScaler
num = ['usia','pendapatan','pendidikan']
sc = StandardScaler()
X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num]) # cegah data leakage

print("X_train setelah scaling:")
print(X_train)

print("\nX_test setelah scaling:")
print(X_test)
```

```python
# Langkah 6 Verifikasi

print("Cek nilai hilang pada X_train:")
print(X_train.isnull().sum())

print("\nCek nilai hilang pada X_test:")
print(X_test.isnull().sum())

print("\nTipe data X_train:")
print(X_train.dtypes)

print("\nTipe data X_test:")
print(X_test.dtypes)

print("\nStatistik X_train:")
print(X_train.describe())

print("\nStatistik X_test:")
print(X_test.describe())
```

```python
# Latihan Mandiri 1

df['usia'] = df['usia'].fillna(df['usia'].mean())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].mean())

print("Mean usia:", df['usia'].mean())
print("Mean usia:", df['usia'].mean())
print("Median usia:", df['usia'].median())
```

```python
# Latihan Mandiri 2

from sklearn.preprocessing import MinMaxScaler

scaler_minmax = MinMaxScaler()

X_train_minmax = X_train.copy()
X_test_minmax = X_test.copy()

num = ['usia', 'pendapatan', 'pendidikan']

X_train_minmax[num] = scaler_minmax.fit_transform(
    X_train_minmax[num]
)

X_test_minmax[num] = scaler_minmax.transform(
    X_test_minmax[num]
)

print("X_train setelah MinMaxScaler:")
print(X_train_minmax.describe())

print("\nX_test setelah MinMaxScaler:")
print(X_test_minmax.describe())
```

```python
# Latihan Mandiri 3

df_latihan = df.copy()

# Menambahkan kolom kategorikal baru
df_latihan['jenis_kelamin'] = 'Laki-laki'

# Mengubah sebagian data menjadi kategori lain
df_latihan.loc[df_latihan.index[::2], 'jenis_kelamin'] = 'Perempuan'

print("Data sebelum one-hot encoding:")
print(df_latihan)

# One-hot encoding
df_latihan = pd.get_dummies(
    df_latihan,
    columns=['jenis_kelamin'],
    dtype=int
)

print("\nData setelah one-hot encoding:")
print(df_latihan)
```

# Latihan Mandiri 4
Pada data latih menggunakan fit_transform karena scaler perlu mempelajari nilai atau parameter dari data latih sekaligus melakukan transformasi, sedangkan data uji menggunakan transform karena data uji tidak boleh digunakan untuk mempelajari parameter scaling, karena untuk mencegah terjadinya data leakage

E. Refleksi
1. Karena data harus dibagi menjadi data latih dan data uji terlebih dahulu sebelum dilakukan scaling, karena scaling dilakukan dengan fit pada data latih saja, kemudian data uji hanya dilakukan transform, karena untuk mencegah terjadinya data leakage
2. Label econding digunakan ketika data kategorikal dapat direpresentasikan  dalam bentuk urutan atau nilai angka, seperti pendidikan SMA,S1, dan S2. Sedangkan one-hot encoding digunakan untuk data kategorikal nominal seperti kota, yang tidak memiliki utrutan tertentu
3. Normalisasi digunakan untuk mengubah nilai fitur ke rentang tertentu, sedangkan standardisasi mengubah data berdasarkan rata-rata dan standar deviasi sehingga skala data menjadi lebih seragam
